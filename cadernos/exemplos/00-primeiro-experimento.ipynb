```python
from pathlib import Path
import sys
import torch
import labia

# O kernel abre com cwd na pasta do caderno; a raiz do laboratório
# é o primeiro ancestral que contém a pasta 'cadernos'.
pasta_atual = Path.cwd().resolve()
raiz = next((p for p in (pasta_atual, *pasta_atual.parents) if (p / 'cadernos').is_dir()), pasta_atual)
resultados = raiz / 'cadernos' / 'resultados'
resultados.mkdir(parents=True, exist_ok=True)
print(f'Lab-IA importado de: {Path(labia.__file__).resolve()}')
print(f'PyTorch: {torch.__version__}; CUDA: {torch.cuda.is_available()}')
```

```python
# Esta variável continua viva para as próximas células desta sessão.
hipotese = 'Uma perda quadrática diminui neste exemplo mínimo.'
amostras = torch.tensor([1.0, 2.0, 3.0])
alvo = torch.tensor([2.0, 4.0, 6.0])
print(hipotese)
```

```python
# Mude a perda, o modelo ou os dados e execute novamente.
peso = torch.nn.Parameter(torch.tensor(0.0))
otimizador = torch.optim.SGD([peso], lr=0.1)
for passo in range(40):
    perda = torch.mean((peso * amostras - alvo) ** 2)
    otimizador.zero_grad()
    perda.backward()
    otimizador.step()
print({'peso': float(peso.detach()), 'perda': float(perda.detach())})
```

```python
# O núcleo do Lab-IA também está disponível como biblioteca.
from labia.models.gpt import ConfigGPT, GPT

torch.manual_seed(7)
config = ConfigGPT(vocab=32, dim=16, camadas=1, cabecas=2, janela_ctx=8, abandono=0.0)
modelo = GPT(config)
modelo.init_pesos(semente=7)
tokens = torch.randint(0, config.vocab, (2, 8))  # troque pelos seus tokens ou dados
logits, perda_gpt = modelo(tokens, tokens)
print({'parâmetros': modelo.contar_parametros(), 'perda': float(perda_gpt.detach())})
```

```python
# Salvar artefatos é explícito; reiniciar o kernel limpa só a memória.
checkpoint = resultados / 'primeiro-experimento.pt'
torch.save({'peso': peso.detach(), 'modelo': modelo.state_dict(), 'config': config.para_dict(), 'hipotese': hipotese}, checkpoint)
restaurado = torch.load(checkpoint, weights_only=True)
print(f'Checkpoint salvo e relido: {checkpoint}')
restaurado
```
